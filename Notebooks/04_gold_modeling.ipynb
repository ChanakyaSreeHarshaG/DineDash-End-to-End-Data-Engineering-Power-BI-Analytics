```python
%sql


CREATE OR REPLACE TABLE gold.dim_customer AS
SELECT *
FROM silver.customers;

CREATE OR REPLACE TABLE gold.dim_restaurant AS
SELECT *
FROM silver.restaurants;

CREATE OR REPLACE TABLE gold.dim_delivery_agent AS
SELECT *
FROM silver.delivery_agents;

CREATE OR REPLACE TABLE gold.dim_location AS
SELECT *
FROM silver.locations;

CREATE OR REPLACE TABLE gold.dim_menu_item AS
SELECT *
FROM silver.menu_items;


-- =========================
-- GOLD FACTS
-- =========================

CREATE OR REPLACE TABLE gold.fact_orders AS
SELECT
    order_id,
    customer_id,
    restaurant_id,
    agent_id,
    delivery_location_id,
    order_status,
    payment_method,
    order_timestamp,
    tip,
    total_amount
FROM silver.orders;


CREATE OR REPLACE TABLE gold.fact_order_items AS
SELECT
    order_id,
    item_id,
    item_name,
    item_price,
    quantity,
    item_price * quantity AS line_total
FROM silver.order_items;
```
